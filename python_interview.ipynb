```python
###Reverse a String

def reverse_string(s):
    rev = ""
    for ch in s:
        rev = ch + rev
    return rev

print(reverse_string("amazon"))
```

```python
####Find Locations of a Word in a Sentence

def find_word_positions(sentence, word):
    words = sentence.split()
    for i in range(len(words)):
        if words[i] == word:
            print(i)

find_word_positions("amazon is big amazon company", "amazon")
```

```python
def word_exists(sentence, word):
    words = sentence.split()
    for w in words:
        if w == word:
            return True
    return False

print(word_exists("AMAZON IS A BIG AMAZON COMPANY", "AMAZON"))
```

```python
####Find Prime Numbers (1 to 100)

def find_primes():
    for num in range(2, 101):
        is_prime = True
        for i in range(2, num):
            if num % i == 0:
                is_prime = False
                break
        if is_prime:
            print(num)

find_primes()
```

```python
###Merge Two Sorted Arrays(Given two sorted integer arrays, merge them into one sorted array.)


def merge_sorted_arrays(a, b):
    result = []
    i = j = 0

    while i < len(a) and j < len(b):
        if a[i] < b[j]:
            result.append(a[i])
            i += 1
        else:
            result.append(b[j])
            j += 1

    result.extend(a[i:])
    result.extend(b[j:])
    return result

print(merge_sorted_arrays([1,3,5], [2,4,6]))
```

```python
####Two Sum Problem (VERY IMPORTANT)
###Given an array of integers and a target number, return the indices of two
###numbers that add up to the target.


def two_sum(nums, target):
    seen = {}

    for i in range(len(nums)):
        needed = target - nums[i]
        if needed in seen:
            return [seen[needed], i]
        seen[nums[i]] = i

print(two_sum([2,7,11,15], 9))
```

```python
###Find All Pairs With Given Sum

###Given a list of integers, find all pairs whose sum equals a given number.


def pair_sum(nums, target):
    pairs = []
    for i in range(len(nums)):
        for j in range(i+1, len(nums)):
            if nums[i] + nums[j] == target:
                pairs.append((nums[i], nums[j]))
    return pairs

print(pair_sum([1,2,3,4,5], 5))
```

```python
####Next Greater Element on Right
###Given an array, replace each element with the next greater element to its right.
###If none exists, replace with -1.

def next_greater(arr):
    result = []

    for i in range(len(arr)):
        next_val = -1
        for j in range(i+1, len(arr)):
            if arr[j] > arr[i]:
                next_val = arr[j]
                break
        result.append(next_val)

    return result

print(next_greater([4,5,2,10]))
```

```python
####Common Elements Between Two Arrays
###Given two string arrays, return all common elements.

def common_elements(a, b):
    result = []
    for item in a:
        if item in b:
            result.append(item)
    return result

print(common_elements(["amazon", "aws"], ["aws", "s3"]))
```

```python
####Count Character Frequency in a String

####Given a string, count how many times each character appears.

def char_frequency(s):
    freq = {}
    for ch in s:
        if ch in freq:
            freq[ch] += 1
        else:
            freq[ch] = 1
    return freq

print(char_frequency("amazon"))
```

```python
####Find Duplicate Elements in a List
###Find duplicate numbers in a list.

def find_duplicates(nums):
    seen = set()
    duplicates = []

    for n in nums:
        if n in seen and n not in duplicates:
            duplicates.append(n)
        else:
            seen.add(n)

    return duplicates

print(find_duplicates([1,2,3,2,4,1]))
```

```python
###Remove Duplicates (Preserve Order)

###Remove duplicates from a list while keeping order.

def remove_duplicates(nums):
    result = []
    for n in nums:
        if n not in result:
            result.append(n)
    return result

print(remove_duplicates([1,2,2,3,1]))
```

```python
####Check if Two Strings Are Anagrams
####Check whether two strings are anagrams.

def is_anagram(s1, s2):
    return sorted(s1) == sorted(s2)

print(is_anagram("listen", "silent"))
```

```python
####First Non-Repeating Character
####Find the first non-repeating character in a string.

def first_unique_char(s):
    for ch in s:
        if s.count(ch) == 1:
            return ch
    return None

print(first_unique_char("amazon"))
```

```python
####Find Missing Number in a Sequence

####Given numbers from 1 to N with one missing, find the missing number.

def missing_number(nums):
    n = len(nums) + 1
    expected_sum = n * (n + 1) // 2
    return expected_sum - sum(nums)

print(missing_number([1,2,3,5]))
```

```python
####Word Count in a Sentence

####Count how many times each word appears in a sentence.

def word_count(sentence):
    words = sentence.split()
    count = {}

    for w in words:
        if w in count:
            count[w] += 1
        else:
            count[w] = 1

    return count

print(word_count("amazon is big amazon company"))
```

```python
####Merge Two Dictionaries
####Merge two dictionaries. If key exists in both, add values.

def merge_dicts(d1, d2):
    result = d1.copy()

    for key in d2:
        if key in result:
            result[key] += d2[key]
        else:
            result[key] = d2[key]

    return result

print(merge_dicts({'a':1,'b':2}, {'b':3,'c':4}))
```

```python
####Read File and Count "ERROR" Lines (QAE REAL)
####Read log lines and count how many contain the word "ERROR".

def count_errors(lines):
    count = 0
    for line in lines:
        if "ERROR" in line:
            count += 1
    return count

logs = [
    "INFO started",
    "ERROR failed",
    "ERROR timeout"
]

print(count_errors(logs))
```
